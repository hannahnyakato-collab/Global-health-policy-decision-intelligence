This notebook imports the World Bank Health, Nutrition and Population dataset and explores its structure before data cleaning and analysis.

```python
import numpy as np
import pandas as pd
```

```python
# pd.set_option ('display.max_columns', None)
# pd.set_option ('display.max_rows', 100)

df = pd.read_excel ("C:\\Users\\786\\Desktop\\document\\Global Health Policy Decision Intelligence\\data\\raw\\world_bank_health_data.xlsx", sheet_name='Data')
df.head()
df.shape
df.columns
df.info()
df.isnull().sum()
```

```python
df.tail(10)
df=df.dropna(subset=['Country Name'])
df.shape
```

```python
df= df.replace('..', np.nan)
```

```python
df.tail()
```

```python
year_cols = [col for col in df.columns if "[YR" in col]
```

```python
df[year_cols] = df[year_cols].apply(pd.to_numeric, errors="coerce")
```

```python
df[year_cols].dtypes.unique()
```

```python
df["Country Name"].nunique()
```

```python
sorted(df["Country Name"].unique())
```

```python
df["Series Name"].nunique()
```

```python
sorted(df["Series Name"].unique())
```

```python
df_long = df.melt(
    id_vars=["Series Name", "Series Code", "Country Name", "Country Code"],
    value_vars=year_cols,
    var_name="Year",
    value_name="Value"
)
```

```python
df_long["Year"] = df_long["Year"].str.extract(r"(\d{4})").astype(int)
```

```python
df_long = df_long.dropna(subset=["Value"])
```
